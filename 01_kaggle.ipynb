# Human trafficking dataset

```python
# Import pandas for working with tabular data.
import pandas as pd

# Load a reproducible sample of 10,000 rows.
df = pd.read_csv("data/human_trafficking.csv").sample(
    n=10000,
    random_state=42,
)
```

```python
# Show the number of rows and columns.
df.shape
```

```python
# Show the data type of every column.
df.dtypes
```

```python
# Show the first five rows.
df.head()
```

```python
# Count missing values in every column.
df.isnull().sum()
```

```python
# Count the two dataset codes used for unknown values.
unknown_counts = pd.DataFrame({
    "-99": (df == -99).sum(),
    "00": (df == "00").sum()
})
unknown_counts
```

```python
# Show counts for registration year.
df["yearOfRegistration"].value_counts()
```

```python
# Show counts for gender.
df["gender"].value_counts()
```

```python
# Show counts for broad age groups.
df["ageBroad"].value_counts()
```

```python
# Show the ten most common citizenship codes.
df["citizenship"].value_counts().head(10)
```

```python
# Show the ten most common exploitation country codes.
df["CountryOfExploitation"].value_counts().head(10)
```

```python
# Show counts for exploitation types.
df["typeOfExploitConcatenated"].value_counts()
```
